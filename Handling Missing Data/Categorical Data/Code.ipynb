### Importing data set

```python
import pandas as pd

url='https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/main/day35-complete-case-analysis/data_science_job.csv'

df=pd.read_csv(url)

df.head()
```

```python
df.isnull().mean()
```

### Train Test Split

```python
from sklearn.model_selection import train_test_split

X=df.drop(columns=['target'], axis=1)
Y=df['target']

X_train, X_test, Y_train, Y_test =train_test_split(X, Y, test_size=0.2, random_state=42)
```

### Column Transformer by using Simple Imputer but with different strategies

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer

trf=ColumnTransformer([
    ('impute_training_hours', SimpleImputer(strategy='mean'), ['training_hours']),
    ('impute_city_development_index', SimpleImputer(strategy='median'), ['city_development_index']),
    ('impute_experience', SimpleImputer(strategy='median'), ['experience']),
    ('impute_enrolled_university', SimpleImputer(strategy='most_frequent'), ['enrolled_university']),
    ('impute_education_level', SimpleImputer(strategy='most_frequent'), ['education_level']),
    ('impute_gender', SimpleImputer(strategy='constant', fill_value='Missing'), ['gender']),  
    ('impute_major_discipline', SimpleImputer(strategy='constant', fill_value='Missing'), ['major_discipline']),
    ('impute_company_size', SimpleImputer(strategy='constant' , fill_value='Missing'), ['company_size']),
    ('impute_company_type', SimpleImputer(strategy='constant' , fill_value='Missing'), ['company_type']),
], remainder='passthrough')
```

### Pipeline

```python
from sklearn.pipeline import Pipeline

pipe=Pipeline([
    ('pre_processing', trf)
])

X_train_transformed=pipe.fit_transform(X_train)
X_test_Transformed=pipe.transform(X_test)
```

### Converting back into Data Frame

```python
import numpy as np

cols=['training_hours', 'city_development_index', 'experience', 'enrolled_university', 'education_level', 'gender', 'major_discipline' , 'company_size', 'company_type'] + [c for c in X_train.columns if c not in ['training_hours', 'city_development_index', 'experience', 'enrolled_university', 'education_level', 'gender', 'major_discipline' ,'company_size', 'company_type']]

final_data_set=pd.DataFrame(X_train_transformed, columns=cols)

final_data_set.head()
```
