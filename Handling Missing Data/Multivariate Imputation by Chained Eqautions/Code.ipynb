### Importing Data Set

```python
import pandas as pd
import numpy as np

url="https://raw.githubusercontent.com/npradaschnor/Pima-Indians-Diabetes-Dataset/master/diabetes.csv"
df=pd.read_csv(url)
df.head()
```

```python
df.isnull().sum()
```

### Data Cleaning by replacing 0 with NaN

```python
columns_to_fix=['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
df[columns_to_fix]=df[columns_to_fix].replace(0, np.nan)

df.head()
```

### Train Test Split

```python
from sklearn.model_selection import train_test_split

X=df.drop(columns=['Outcome'], axis=1)
Y=df['Outcome']

X_train, X_test, Y_train, Y_test= train_test_split(X, Y, test_size=0.2, random_state=42)
```

### Random Forest Regression

```python
from sklearn.ensemble import RandomForestRegressor
 
rf_brain=RandomForestRegressor(n_estimators=10, random_state=42)
```

### Iterative Imputer

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

imputer=IterativeImputer(
    estimator=rf_brain,
    random_state=42,
    verbose=2
)
```

### Fit and Transform

```python
X_train_imputed=imputer.fit_transform(X_train)
X_test_imputed=imputer.transform(X_test)
```

### Converting back to Data Frame

```python
X_train_final = pd.DataFrame(X_train_imputed, columns=X.columns)
X_test_final = pd.DataFrame(X_test_imputed, columns=X.columns)

# 8. RESULTS CHECK
print("--- Training Data Missing Values ---")
print(X_train_final.isnull().sum())

print("\n--- Testing Data Missing Values ---")
print(X_test_final.isnull().sum())

display(X_train_final.head())
```
