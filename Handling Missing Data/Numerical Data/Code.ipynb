### Importing Data Set

```python
import pandas as pd

url='https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/main/day35-complete-case-analysis/data_science_job.csv'

df=pd.read_csv(url)

df.head()
```

### Train Test Split

```python
from sklearn.model_selection import train_test_split

X=df.drop(columns=['target'], axis=1)
Y=df['target']

X_train, X_test, Y_train, Y_test=train_test_split(X,Y,test_size=0.2, random_state=42)
```

### Colum Transformer By using Simple Imputer

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer

trf=ColumnTransformer([
    ('imputer_city', SimpleImputer(strategy='mean'), ['city_development_index']),
    ('impute_hours', SimpleImputer(strategy='median'), ['training_hours']),
    ('impute_exp', SimpleImputer(strategy='constant', fill_value=99), ['experience'])
], remainder='passthrough')
```

### Pipeline

```python
import numpy as np
from sklearn.pipeline import Pipeline

pipe=Pipeline([
    ('pre_processing', trf)
])

X_train_transformed=pipe.fit_transform(X_train)
X_test_transformed=pipe.transform(X_test)
```

### Converting into DataFrame

```python
cols=['city_development_index', 'training_hours', 'experience'] + [c for c in X_train.columns if c not in ['city_development_index', 'training_hours', 'experience']]
df_final=pd.DataFrame(X_train_transformed, columns=cols)

df_final.head()
```
