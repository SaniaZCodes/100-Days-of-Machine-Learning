```python
import pandas as pd
import numpy as np

# This is the Diabetes dataset with proper headers
url = 'https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.csv'
cols = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI', 'DiabetesPedigree', 'Age', 'Outcome']
df = pd.read_csv(url, names=cols)

# IMPORTANT: In this dataset, 0 means the value is missing (except for Pregnancies/Outcome)
# We must replace 0 with NaN so the KNNImputer can see them
cols_to_fix = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
df[cols_to_fix] = df[cols_to_fix].replace(0, np.nan)
df.head()
#print(df.isnull().sum())
```

### Train Test Split

```python
from sklearn.model_selection import train_test_split

X=df.drop(columns=['Outcome'], axis=1)
Y= df['Outcome']

X_train, X_test, Y_train, Y_test= train_test_split(X, Y, test_size=0.2, random_state=42)
```

### Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()

X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

### KNN Imputer

```python
from sklearn.impute import KNNImputer

knn= KNNImputer(n_neighbors=5, weights='distance')

X_train_imputed=knn.fit_transform(X_train_scaled)
X_test_imputed=knn.transform(X_test_scaled)
```

### Converting to Data Frame

```python
X_train_final=pd.DataFrame(X_train_imputed, columns=X.columns)
X_train_final.isnull().sum()
```
