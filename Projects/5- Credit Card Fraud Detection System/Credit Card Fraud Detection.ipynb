# Step 4: Train Test Split

```python
# Features and Target
X = df.drop('Class', axis=1)
Y = df['Class']

# Train-Test Split
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42,
    stratify=Y
)

# Check Shapes
print("X_train Shape:", X_train.shape)
print("X_test Shape :", X_test.shape)

print("Y_train Shape:", Y_train.shape)
print("Y_test Shape :", Y_test.shape)

# Check Class Distribution
print("\nTraining Set Distribution:")
print(Y_train.value_counts())

print("\nTesting Set Distribution:")
print(Y_test.value_counts())

# Check Percentage Distribution
print("\nTraining Set Percentage:")
print(Y_train.value_counts(normalize=True) * 100)

print("\nTesting Set Percentage:")
print(Y_test.value_counts(normalize=True) * 100)
```

# Step 6: Transformation

```python
from sklearn.preprocessing import PowerTransformer

# Create transformer
pt = PowerTransformer(method='yeo-johnson')

# Apply only on Amount
X_train['Amount'] = pt.fit_transform(
    X_train[['Amount']]
)

X_test['Amount'] = pt.transform(
    X_test[['Amount']]
)
```

```python
X_train['Amount'].skew()
```

# Step 7: Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Fit on training data
X_train = pd.DataFrame(
    scaler.fit_transform(X_train),
    columns=X_train.columns,
    index=X_train.index
)

# Transform test data
X_test = pd.DataFrame(
    scaler.transform(X_test),
    columns=X_test.columns,
    index=X_test.index
)
```

```python
X_train.describe().round(2)
```

# Step 8: SMOTE

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(
    random_state=42
)

X_train_smote, Y_train_smote = smote.fit_resample(
    X_train,
    Y_train
)
```

```python
import sklearn
import imblearn

print("Scikit-Learn:", sklearn.__version__)
print("Imbalanced-Learn:", imblearn.__version__)
```

```python
print("Before SMOTE:")
print(Y_train.value_counts())

print("\nAfter SMOTE:")
print(Y_train_smote.value_counts())
```

# Step 9: Model

## Random Forest Classifier

```python
from sklearn.ensemble import RandomForestClassifier

# Create Model
rf = RandomForestClassifier(
    n_estimators=30,
    max_depth=10,
    random_state=42,
    n_jobs=-1
)

# Train Model
rf.fit(
    X_train_smote,
    Y_train_smote
)

# Predictions
y_pred_rf = rf.predict(X_test)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

rf_accuracy = accuracy_score(Y_test, y_pred_rf)

rf_precision = precision_score(Y_test, y_pred_rf)

rf_recall = recall_score(Y_test, y_pred_rf)

rf_f1 = f1_score(Y_test, y_pred_rf)

y_prob_rf = rf.predict_proba(X_test)[:,1]

rf_auc = roc_auc_score(
    Y_test,
    y_prob_rf
)

print("Accuracy :", rf_accuracy)
print("Precision :", rf_precision)
print("Recall :", rf_recall)
print("F1 Score :", rf_f1)
print("ROC-AUC :", rf_auc)
```

## Stacking Classifier

```python
from sklearn.ensemble import (
    StackingClassifier,
    RandomForestClassifier,
    GradientBoostingClassifier
)

from sklearn.linear_model import LogisticRegression

# Base Models
base_models = [
    (
        'rf',
        RandomForestClassifier(
            n_estimators=20,
            max_depth=8,
            random_state=42,
            n_jobs=-1
        )
    ),

    (
        'gb',
        GradientBoostingClassifier(
            n_estimators=20,
            random_state=42
        )
    )
]

# Stacking Model
stack = StackingClassifier(
    estimators=base_models,
    final_estimator=LogisticRegression()
)

# Train Model
stack.fit(
    X_train_smote,
    Y_train_smote
)

# Predictions
y_pred_stack = stack.predict(X_test)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

stack_accuracy = accuracy_score(
    Y_test,
    y_pred_stack
)

stack_precision = precision_score(
    Y_test,
    y_pred_stack
)

stack_recall = recall_score(
    Y_test,
    y_pred_stack
)

stack_f1 = f1_score(
    Y_test,
    y_pred_stack
)

y_prob_stack = stack.predict_proba(X_test)[:,1]

stack_auc = roc_auc_score(
    Y_test,
    y_prob_stack
)

print("Accuracy :", stack_accuracy)
print("Precision :", stack_precision)
print("Recall :", stack_recall)
print("F1 Score :", stack_f1)
print("ROC-AUC :", stack_auc)
```

# Step 11: Optuna Tuning on Random Forest

## Objective Function

```python
import optuna

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score

# Objective Function
def objective(trial):

    model = RandomForestClassifier(
        n_estimators=trial.suggest_int(
            "n_estimators",
            50,
            300
        ),

        max_depth=trial.suggest_int(
            "max_depth",
            3,
            20
        ),

        min_samples_split=trial.suggest_int(
            "min_samples_split",
            2,
            20
        ),

        min_samples_leaf=trial.suggest_int(
            "min_samples_leaf",
            1,
            10
        ),

        random_state=42,
        n_jobs=-1
    )

    score = cross_val_score(
        model,
        X_train_smote,
        Y_train_smote,
        cv=3,
        scoring="roc_auc",
        n_jobs=-1
    ).mean()

    return score


# Create Study
study = optuna.create_study(
    direction="maximize"
)

# Start Optimization
study.optimize(
    objective,
    n_trials=10
)

# Best Results
print("Best Parameters:")
print(study.best_params)

print("\nBest ROC-AUC:")
print(study.best_value)
```
