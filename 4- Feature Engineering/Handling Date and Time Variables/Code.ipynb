### Importing Data Set

```python
import pandas as pd
import numpy as np

# Creating a manual version of the Uber dataset for practice
data = {
    'START_DATE*': ['1/1/2016 21:11', '1/2/2016 1:25', '1/2/2016 20:25', '1/5/2016 17:31', '1/6/2016 14:42', '10/25/2016 22:15'],
    'END_DATE*': ['1/1/2016 21:17', '1/2/2016 1:37', '1/2/2016 20:38', '1/5/2016 17:45', '1/6/2016 15:49', '10/26/2016 01:10'],
    'CATEGORY*': ['Business', 'Business', 'Business', 'Business', 'Business', 'Personal'],
    'MILES*': [5.1, 5.0, 4.8, 4.7, 63.7, 12.5],
    'PURPOSE*': ['Meal/Entertain', 'Errand/Supplies', 'Meeting', 'Customer Visit', 'Meal/Entertain', 'Medical']
}

df = pd.DataFrame(data)
print("Data Created Locally! No more URL errors.")
df.head()
```

```python
df.info()
```

```python
df['START_DATE*']=pd.to_datetime(df['START_DATE*'])
df['END_DATE*']=pd.to_datetime(df['END_DATE*'])
```

### Extracting Basic Features

```python
df['year']=df['START_DATE*'].dt.year
df['month_name']=df['START_DATE*'].dt.month_name()
df['day_name']=df['START_DATE*'].dt.day_name()
df['hour']=df['START_DATE*'].dt.hour

df['duration']=df['END_DATE*']- df['START_DATE*']

df['is_weekend']=np.where(df['START_DATE*'].dt.dayofweek>4, 1,0)

print("Data types after conversion")
print(df.dtypes)
df.head()
```

```python
df['duration_minutes'] = df['duration'].dt.total_seconds() / 60
df.head()
```
