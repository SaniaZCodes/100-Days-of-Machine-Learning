### Importing Data Set

```python
import pandas as pd
import numpy as np

df=pd.read_csv('train.csv')

df.head()
```

### Feature Construction

```python
df['family_size']=df['SibSp']+df['Parch']+1

print(df['family_size'])
```

### Feature Splitting

```python
df['title']= df['Name'].str.split(',').str.get(1).str.split('.').str.get(0).str.strip()

print(df['title'])
```

### Dropping Extra Columns

```python
df.drop(columns=['SibSp', 'Parch','Name'], inplace=True)
```
