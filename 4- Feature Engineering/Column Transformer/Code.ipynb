```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder, LabelEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
# 1. Load Data (Keeping 'gender' this time)
df = pd.read_csv('https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/main/day26-ordinal-encoding/customer.csv')
df = df[['age', 'gender', 'education', 'purchased']]
```

```python
# 2. Split Data
X_train, X_test, y_train, y_test = train_test_split(df.drop('purchased', axis=1), 
                                                    df['purchased'], 
                                                    test_size=0.2, 
                                                    random_state=42)
```

```python
# 3. Create the ColumnTransformer
# Syntax: ('name', encoder_object, [column_indices_or_names])
transformer = ColumnTransformer(transformers=[
    ('tnf1', OrdinalEncoder(categories=[['School', 'UG', 'PG']]), ['education']),
    ('tnf2', OneHotEncoder(drop='first', sparse_output=False), ['gender'])
], remainder='passthrough') # 'passthrough' tells it NOT to delete the 'age' column!
```

```python
# 4. Fit and Transform
# Note: ColumnTransformer returns a NumPy array, so we wrap it back in a DataFrame to see it clearly
X_train_transformed = transformer.fit_transform(X_train)
X_test_transformed = transformer.transform(X_test)
```

```python
# 5. Label Encode the Target (y)
le = LabelEncoder()
y_train = le.fit_transform(y_train)
y_test = le.transform(y_test)
```

```python
# 6. Verification
# Convert to DataFrame just for printing
X_train_transformed = pd.DataFrame(X_train_transformed, columns=['education', 'gender', 'age'])
print("--- Fully Encoded Data ---")
print(X_train_transformed.head())
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

# 1. Load the New Dataset (Car Details)
df = pd.read_csv('https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/main/day27-one-hot-encoding/cars.csv')

# We'll take a subset of columns to keep it clean for practice
df = df[['fuel', 'owner', 'km_driven', 'selling_price']]

# 2. Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(df.drop('selling_price', axis=1),
                                                    df['selling_price'],
                                                    test_size=0.2,
                                                    random_state=42)

# 3. Create ColumnTransformer for One-Hot Encoding
# We are encoding 'fuel' and 'owner'
transformer = ColumnTransformer(transformers=[
    ('tnf_ohe', OneHotEncoder(drop='first', sparse_output=False), ['fuel', 'owner'])
], remainder='passthrough')

# 4. Fit and Transform
X_train_transformed = transformer.fit_transform(X_train)
X_test_transformed = transformer.transform(X_test)

# 5. Verification
# Let's see how the columns "exploded"
print("Shape of X_train before:", X_train.shape)
print("Shape of X_train after OHE:", X_train_transformed.shape)

# Convert back to DF to see the 0s and 1s
columns = transformer.get_feature_names_out()
X_train_final = pd.DataFrame(X_train_transformed, columns=columns)

print("\n--- Encoded Data (First 5 rows) ---")
print(X_train_final.head())
```
