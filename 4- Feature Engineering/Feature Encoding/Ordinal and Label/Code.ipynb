```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder, LabelEncoder
```

```python
# 1. Load the Dataset
df = pd.read_csv('https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/main/day26-ordinal-encoding/customer.csv')

# We only keep 'age', 'education' (Ordinal), and 'purchased' (Target)
df = df[['age', 'education', 'purchased']]
```

```python
# 2. Train-Test Split (Standard Procedure)
X_train, X_test, y_train, y_test = train_test_split(df.drop('purchased', axis=1), 
                                                    df['purchased'], 
                                                    test_size=0.2, 
                                                    random_state=42)
```

```python
# --- STEP 3: ORDINAL ENCODING ---
oe = OrdinalEncoder(categories=[['School', 'UG', 'PG']])

# IMPORTANT: You must ASSIGN (=) the result back to the column!
X_train['education'] = oe.fit_transform(X_train[['education']])
X_test['education'] = oe.transform(X_test[['education']])
```

```python
# --- STEP 4: LABEL ENCODING ---
le = LabelEncoder()
y_train = le.fit_transform(y_train)
y_test = le.transform(y_test)
```
