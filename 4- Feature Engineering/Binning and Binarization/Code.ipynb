### Importing data set

```python
import pandas as pd
import numpy as np

data = {
    'Age': [22, 45, 30, 50, 24, 60, 40, 33, 29, 55, 19, 21, 44, 37, 52],
    'Annual_Income': [30000, 100000, 45000, 120000, 35000, 150000, 80000, 60000, 42000, 110000, 25000, 28000, 95000, 72000, 105000],
    'Spending_Score': [15, 80, 35, 90, 20, 95, 50, 45, 30, 85, 10, 12, 75, 55, 88],
    'City': ['New York', 'London', 'London', 'Paris', 'Paris', 'New York', 'London', 'Paris', 'New York', 'London', 'Paris', 'New York', 'London', 'Paris', 'New York'],
    'Membership': ['Normal', 'Premium', 'Normal', 'Premium', 'Normal', 'Premium', 'Normal', 'Normal', 'Normal', 'Premium', 'Normal', 'Normal', 'Premium', 'Normal', 'Premium']
}

df = pd.DataFrame(data)
df.to_csv('practice_data.csv', index=False)
print("Dataset created successfully as 'practice_data.csv'!")
```

### Train Test Split

```python
from sklearn.model_selection import train_test_split

X=df.drop('Membership', axis=1)
Y=df['Membership']

X_train, X_test, Y_train, Y_test=train_test_split(X, Y, test_size=0.2, random_state=42)
```

### Feature Scaling

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import KBinsDiscretizer, Binarizer, OneHotEncoder

trf= ColumnTransformer(transformers=[
    ('income_bining', KBinsDiscretizer(n_bins=3, strategy='quantile', encode='ordinal'), [1]),
    ('spending_binarizer', Binarizer(threshold=50), [2]),
    ('city_encode', OneHotEncoder(drop='first', sparse_output=False), [3])
], remainder='passthrough')
```

```python
from sklearn.pipeline import Pipeline

pipe=Pipeline([
    ('preprocessing', trf)
])

X_train_transformed=pipe.fit_transform(X_train)
X_test_transformed=pipe.transform(X_test)

final_columns=['Income_Bin', 'High_Spender', 'City_NY', 'City_Paris', 'Age']

X_train_final=pd.DataFrame(X_train_transformed, columns= final_columns)
X_test_final=pd.DataFrame(X_test_transformed, columns= final_columns)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import KBinsDiscretizer, Binarizer, OneHotEncoder
from sklearn.pipeline import Pipeline
import pandas as pd

# 1. Update the Transformer to use 'kmeans'
trf_kmeans = ColumnTransformer([
    ('income_binning', KBinsDiscretizer(n_bins=3, strategy='kmeans', encode='ordinal', random_state=42), [1]),
    ('spending_binarize', Binarizer(threshold=50), [2]),
    ('city_encode', OneHotEncoder(drop='first', sparse_output=False), [3])
], remainder='passthrough')

# 2. Create and Run the Pipeline
pipe_kmeans = Pipeline([
    ('preprocessing', trf_kmeans)
])

X_train_transformed = pipe_kmeans.fit_transform(X_train)

# 3. Convert to DataFrame
final_columns = ['Income_Bin', 'High_Spender', 'City_NY', 'City_Paris', 'Age']
X_train_final_kmeans = pd.DataFrame(X_train_transformed, columns=final_columns)

# 4. Show the result
print("K-Means Binning Results:")
X_train_final_kmeans.head(10)
```

```python
# 1. Reach into the ColumnTransformer to get the binning object
# We use the name we gave it: 'income_binning'
binning_tool = trf_kmeans.named_transformers_['income_binning']

# 2. Print the edges
print("K-Means Salary Ranges:")
print(binning_tool.bin_edges_)
```
