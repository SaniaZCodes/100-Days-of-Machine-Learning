```python
import seaborn as sns
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import RobustScaler
```

```python
# 1. Load Data
df = sns.load_dataset('tips')
X = df[['total_bill', 'tip']]
X
```

```python
# 2. Train-Test Split (The Exam Setup)
X_train, X_test = train_test_split(X, test_size=0.2, random_state=42)
```

```python
# 3. Initialize the "Bodyguard" (RobustScaler)
# This ignores the 'crazy' outliers and focuses on the middle 50%
scaler = RobustScaler()
```

```python
# 4. Fit & Transform (Learning from Train, Applying to both)
# Shortcut: fit_transform for training data
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# 5. Convert back to DataFrame for verification
X_train_scaled = pd.DataFrame(X_train_scaled, columns=X_train.columns)
```

```python
# 6. Final Verification Output
print("--- First 5 Rows of Robust Scaled Data ---")
print(X_train_scaled.head())

print("\n--- The Robust Check ---")
# In Robust Scaling, the MEDIAN becomes 0 (not the mean!)
print(f"Median (Should be 0.0): \n{X_train_scaled.median().round(2)}")

# The Interquartile Range (IQR) after scaling becomes 1.0
# We check this by subtracting Q1 (25%) from Q3 (75%)
iqr = X_train_scaled.quantile(0.75) - X_train_scaled.quantile(0.25)
print(f"\nIQR (Should be 1.0): \n{iqr.round(2)}")
```

### MinMax Scaling

```python
import seaborn as sns
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
```

```python
# 1. Load and Prepare Data
df = sns.load_dataset('tips')
X = df[['total_bill', 'tip']]
```

```python
# 2. Split (Always before scaling!)
X_train, X_test = train_test_split(X, test_size=0.2, random_state=42)
```

```python
# 3. Initialize "The Box" (MinMaxScaler)
# feature_range=(0, 1) is the default
scaler = MinMaxScaler()
```

```python
# 4. Fit & Transform
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# 5. Convert back to DataFrame
X_train_scaled = pd.DataFrame(X_train_scaled, columns=X_train.columns)
```

```python
# 6. Final Verification Output
print("--- Min-Max Scaled Data (First 5 rows) ---")
print(X_train_scaled.head())

print("\n--- The Box Check ---")
# The Minimum MUST be 0 and the Maximum MUST be 1
print(f"Minimum Value: \n{X_train_scaled.min()}")
print(f"\nMaximum Value: \n{X_train_scaled.max()}")
```
