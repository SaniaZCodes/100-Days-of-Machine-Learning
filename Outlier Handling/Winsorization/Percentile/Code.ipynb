### Importing Data Set

```python
import pandas as pd
import numpy as np

df=pd.read_csv('concrete.csv')

df.head()
```

```python
col='strength'

upper_limit=df[col].quantile(0.88)
lower_limit=df[col].quantile(0.22)

max_before=df[col].max()
min_before=df[col].min()

df[col]=np.where(
    df[col]>upper_limit, upper_limit,
    np.where(df[col]<lower_limit, lower_limit, df[col])
)

max_after=df[col].max()
min_after=df[col].min()

print(f""" 
        Upper Limit: {upper_limit:.2f}
        Lower Limit: {lower_limit:.2f}
        ---
        Before capping:
        Max: {max_before:.2f}
        Min: {min_before:.2f}
        ---
        After capping:
        Max: {max_after:.2f}
        Min: {min_after:.2f}
        """)
```
