### Importing Data Set

```python
import seaborn as sns
import pandas as pd
import numpy as np

df= sns.load_dataset('tips')
```

```python
df.head()
```

### Train Test Split

```python
from sklearn.model_selection import train_test_split

X=df.drop(columns=['tip'], axis=1)
Y=df['tip']

X_train, X_test, Y_train, Y_test=train_test_split(X, Y, test_size=0.2, random_state=42)
```

### Finding Sqeweness

```python
columns_to_work_on=['total_bill', 'tip']
for x in columns_to_work_on:
    skew_values=df[x].skew()
    print(f"{x}: {skew_values:.2f}")
```

### Applying IQR Technique

## 'we donot cap Y (target)'

```python
column=['total_bill']
for x in column:

    # finding percentiles
    
    Q1=X_train[x].quantile(0.25)
    Q3=X_train[x].quantile(0.75)

    # finding IQR range

    IQR=Q3-Q1

    # finding limits

    upper_limit=Q3 + 1.5*IQR
    lower_limit=Q1 - 1.5*IQR

    # Calculate before max, min value

    max_before=X_train[x].max()
    min_before=X_train[x].min()

    # Applying Capping to Trainig, Testing (X) data set

    X_train[x]=np.where(
        X_train[x]>upper_limit, upper_limit,
        np.where(
            X_train[x]<lower_limit, lower_limit,
            X_train[x]
        )
    )

    X_test[x]=np.where(
        X_test[x]>upper_limit, upper_limit,
        np.where(
            X_test[x]<lower_limit, lower_limit,
            X_test[x]
        )
    )


    # Calculate after max, min value

    max_after=X_train[x].max()
    min_after=X_train[x].min()


    # printing values for our information

    print(f"""
                Column: {x}
                Q1: {Q1:.2f}
                Q3: {Q3:.2f}
                IQR: {IQR:.2f}
                Upper Limit: {upper_limit:.2f}
                Lower Limit: {lower_limit:.2f}
                Maximum: 
                before: {max_before:.2f} | after: {max_after:.2f}
                Minimum:
                before: {min_before:.2f} | after: {min_after:.2f}
                
    """)
```
